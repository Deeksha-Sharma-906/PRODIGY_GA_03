## Implement a simple text generation algorithm using Markov chains. This task involves creating a statistical model that predicts the probability of a character or word based on the previous one(s).

```python
import markovify
import re
```

```python
def clean_text(text):
    """Basic text cleaning: remove special characters and extra spaces."""
    text = re.sub(r'\s+', ' ', text)  # remove extra spaces
    text = re.sub(r'[^A-Za-z0-9.,!?\'\" ]+', '', text)  # keep only common characters
    return text
```

```python
def load_text(file_path):
    """Load and clean text from a file."""
    with open(file_path, 'r', encoding='utf-8') as file:
        raw_text = file.read()
    return clean_text(raw_text)
```

```python
def build_markov_model(text, state_size=2):
    """Build a Markov model using markovify."""
    return markovify.Text(text, state_size=state_size)
```

```python
def generate_sentences(model, num_sentences=5):
    """Generate sentences from the Markov model."""
    for _ in range(num_sentences):
        print(model.make_sentence())
```

```python
def main():
    # Load your input text file
    file_path = '1342-0.txt'  # Replace with your own text file
    text = load_text(file_path)

    # Build the Markov model
    markov_model = build_markov_model(text, state_size=2)

    # Generate sentences
    print("\nGenerated Sentences:\n")
    generate_sentences(markov_model, num_sentences=5)

if __name__ == "__main__":
    main()
```
